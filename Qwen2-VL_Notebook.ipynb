    <h2 style="color: rgb(247, 230, 202); text-align: center; font-size: 150%; font-weight: bold;">Cook Your Data So That LLM Can Eat</h2>
    <ul style="list-style-type: square; padding-left: 20px;">
        <li><strong>Function Overview</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>Processes a pandas DataFrame to transform it into a structured format.</li>
                <li>Extracts specific columns, formats the data, and saves the results in a JSONL file.</li>
            </ul>
        </li>
        <li style="margin-top: 20px;"><strong>Key Steps</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li><strong>Input Selection</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Extracts <code>prompt</code>, <code>response_a</code>, and <code>response_b</code> as input features.</li>
                        <li>Extracts <code>winner</code> as the target labels.</li>
                    </ul>
                </li>
                <li><strong>Data Formatting</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Iterates over each row in the DataFrame.</li>
                        <li>Combines the prompt and model responses into a single text string.</li>
                        <li>Creates a dictionary with <code>instruction</code>, <code>input</code>, and <code>output</code>.</li>
                    </ul>
                </li>
                <li><strong>Output Generation</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Writes each formatted message to a JSONL file (one JSON object per line).</li>
                    </ul>
                </li>
            </ul>
        </li>
        <li style="margin-top: 20px;"><strong>Use Case</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>Useful for preparing datasets for tasks like model comparison, preference learning, or instruction-based fine-tuning.</li>
            </ul>
        </li>
    </ul>
</div>

```python
def data_transfer(dataframe, output_pt, instruction):
    # Select columns 'prompt', 'response_a', and 'response_b' as input features (x).
    x_cols = ['prompt', 'response_a', 'response_b']
    x = dataframe[x_cols]

    # Select the 'winner' column as the target labels (y).
    y = dataframe['winner']

    # Initialize an empty list to store formatted messages.
    messages = []

    # Iterate over each row in the input dataframe.
    for idx in range(len(x)):
        # Extract the current row's data.
        x_line = x.iloc[idx, :]

        # Format the prompt and model responses.
        prompt = 'PROMPT: ' + x_line['prompt']
        model_a = 'MODEL_A: ' + x_line['response_a']
        model_b = 'MODEL_B: ' + x_line['response_b']

        # Combine the prompt and model responses into a single text.
        text = prompt + model_a + model_b

        # Get the target label for the current row.
        target = y[idx]

        # Create a dictionary representing the message with instruction, input, and output.
        message = {
            'instruction': instruction,  # Instruction provided as a function argument.
            'input': text,               # Combined text of prompt and model responses.
            'output': target             # Target label (winner).
        }

        # Append the message to the list of messages.
        messages.append(message)

    # Write the messages to a JSONL file (one JSON object per line).
    with open(output_pt, "w", encoding="utf-8") as file:
        for message in messages:
            # Write each message as a JSON object, ensuring non-ASCII characters are preserved.
            file.write(json.dumps(message, ensure_ascii=False) + "\n")
```

    <h2 style="color: rgb(247, 230, 202); text-align: center; font-size: 150%; font-weight: bold;">Tokenize it!</h2>
    <ul style="list-style-type: square; padding-left: 20px;">
        <li><strong>Function Overview</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>Processes a single example into a format suitable for training sequence-to-sequence models.</li>
                <li>Tokenizes input and output, combines them, and ensures the sequence length does not exceed a maximum.</li>
            </ul>
        </li>
        <li style="margin-top: 20px;"><strong>Key Steps</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li><strong>Tokenization</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Tokenizes the instruction and input using a structured format (system, user, and assistant roles).</li>
                        <li>Tokenizes the output (response) separately.</li>
                    </ul>
                </li>
                <li><strong>Sequence Construction</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Combines the tokenized instruction and response into a single sequence.</li>
                        <li>Appends a padding token to the end of the sequence.</li>
                    </ul>
                </li>
                <li><strong>Attention Mask and Labels</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Constructs an attention mask to indicate padding and non-padding tokens.</li>
                        <li>Creates labels, masking the instruction part with <code>-100</code> (ignored during loss calculation).</li>
                    </ul>
                </li>
                <li><strong>Truncation</strong>:
                    <ul style="list-style-type: disc; margin-top: 10px; margin-bottom: 10px;">
                        <li>Truncates the sequence if it exceeds the maximum length.</li>
                    </ul>
                </li>
            </ul>
        </li>
        <li style="margin-top: 20px;"><strong>Use Case</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>Used in preprocessing pipelines for fine-tuning language models on instruction-following tasks.</li>
            </ul>
        </li>
    </ul>
</div>

<div style="background-color: rgb(74, 74, 74); border: 4px solid rgb(247, 230, 202); border-radius: 40px; padding: 20px; font-family: 'Roboto'; color: rgb(255, 182, 193); text-align: left; font-size: 120%;">
    <h2 style="color: rgb(255, 182, 193); text-align: center; font-size: 150%; font-weight: bold;">Note on Sequence Truncation</h2>
    <ul style="list-style-type: square; padding-left: 20px;">
        <li><strong>Current Implementation</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>To prevent CUDA out-of-memory errors, the tokenized sequence is truncated if it exceeds the maximum length (<code>MAX_LENGTH</code>).</li>
                <li>The truncation is applied to the end of the combined sequence (including the prompt, Model A's response, and Model B's response).</li>
            </ul>
        </li>
        <li style="margin-top: 20px;"><strong>Potential Issue</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>Directly truncating the end of the sequence may unfairly impact Model B's response, as it is more likely to be cut off.</li>
            </ul>
        </li>
        <li style="margin-top: 20px;"><strong>Proposed Improvement</strong>:
            <ul style="list-style-type: circle; margin-top: 10px; margin-bottom: 10px;">
                <li>Thanks 星瞳's suggestion.<a href="https://www.kaggle.com/csuer411" style="color: rgb(255, 182, 193); text-decoration: underline;">csuer411 on Kaggle</a></li>
                <li>Instead of truncating the end, randomly remove some words from the sequence to control its length.</li>
                <li>This approach is expected to improve model accuracy by preserving more balanced information from both models' responses.</li>
                <li><em>Note</em>: This method has not been tested yet and is based on theoretical considerations.</li>
            </ul>
        </li>
    </ul>
</div>

```python
def process_func(example):
    # Define a maximum length for truncation to avoid excessively long sequences.
    MAX_LENGTH = 5000

    # Initialize empty lists to store tokenized input IDs, attention masks, and labels.
    input_ids, attention_mask, labels = [], [], []

    # Tokenize the instruction and input using the tokenizer.
    # The format includes system, user, and assistant roles for structured input.
    instruction = tokenizer(
        f"<|im_start|>system\n{example['instruction']}<|im_end|>\n<|im_start|>user\n{example['input']}<|im_end|>\n<|im_start|>assistant\n",
        add_special_tokens=False  # Do not add special tokens (e.g., BOS, EOS) automatically.
    )

    # Tokenize the output (response) using the tokenizer.
    response = tokenizer(
        f"{example['output']}",
        add_special_tokens=False  # Do not add special tokens automatically.
    )

    # Combine the input IDs from the instruction and response, and append a padding token.
    input_ids = instruction["input_ids"] + response["input_ids"] + [tokenizer.pad_token_id]

    # Combine the attention masks from the instruction and response, and append a 1 (indicating non-padding).
    attention_mask = (
        instruction["attention_mask"] + response["attention_mask"] + [1]
    )

    # Create labels for the model. Mask the instruction part with -100 (ignored during loss calculation).
    labels = [-100] * len(instruction["input_ids"]) + response["input_ids"] + [tokenizer.pad_token_id]

    # Truncate the sequences if they exceed the maximum length.
    if len(input_ids) > MAX_LENGTH:
        input_ids = input_ids[:MAX_LENGTH]
        attention_mask = attention_mask[:MAX_LENGTH]
        labels = labels[:MAX_LENGTH]

    # Return the processed data as a dictionary.
    return {
        "input_ids": input_ids,  # Token IDs for the input sequence.
        "attention_mask": attention_mask,  # Attention mask for the input sequence.
        "labels": labels  # Labels for the model (with instruction part masked).
    }
```

```python
# Define the path to the original dataset file in parquet format.
original_pt = "dataset/train.parquet"
# Load the original dataset into a pandas DataFrame.
original = pd.read_parquet(original_pt)
# Split the dataset into training and testing sets with a 99% training size.
train, test = train_test_split(original, train_size=0.99)
# Reset the index of the training set to ensure it starts from 0 and is continuous.
train.index = [i for i in range(train.shape[0])]
# Reset the index of the testing set to ensure it starts from 0 and is continuous.
test.index = [i for i in range(test.shape[0])]
```

```python
# Define the output file paths for the processed training and testing datasets.
train_pt = "./cooked-dataset/train.jsonl"
test_pt = "./cooked-dataset/test.jsonl"

# Define the instruction text that will be used to guide the model during training.
instruction = """In the text provided below, PROMPT is the question presented; MODEL_A is the response from the first model; MODEL_B is the response from the second model. Please select the best answer from the two responses above. If the first answer is better, return "model_a"; if the second answer is better, return "model_b"."""

# Process the training dataset and save it to the specified output file.
data_transfer(train, train_pt, instruction)

# Process the testing dataset and save it to the specified output file.
data_transfer(test, test_pt, instruction)
```

## load model

```python
# Define the path to the pre-trained model.
model_pt = "/kaggle/input/qwen_0.5b_instruct/transformers/default/1"

# Load the tokenizer for the pre-trained model.
# `use_fast=False` disables the fast tokenizer, and `trust_remote_code=True` allows executing remote code for custom tokenizers.
tokenizer = AutoTokenizer.from_pretrained(model_pt, use_fast=False, trust_remote_code=True)

# Load the pre-trained causal language model.
# `device_map="auto"` automatically assigns the model to available devices (e.g., GPU).
# `torch_dtype=torch.bfloat16` sets the model's data type to bfloat16 for memory efficiency.
model = AutoModelForCausalLM.from_pretrained(model_pt, device_map="auto", torch_dtype=torch.bfloat16)

# Enable gradient computation for the model's input embeddings.
model.enable_input_require_grads()
```

```python
# Load the training dataset from a JSONL file into a pandas DataFrame.
train_df = pd.read_json('cooked-dataset/train.jsonl', lines=True)

# Convert the pandas DataFrame into a Hugging Face Dataset.
train_ds = Dataset.from_pandas(train_df)

# Apply the `process_func` function to preprocess the training dataset and remove original columns.
train_dataset = train_ds.map(process_func, remove_columns=train_ds.column_names)
```

```python
# Load the testing dataset from a JSONL file into a pandas DataFrame.
test_df = pd.read_json('cooked-dataset/test.jsonl', lines=True)

# Convert the pandas DataFrame into a Hugging Face Dataset.
test_ds = Dataset.from_pandas(test_df)

# Apply the `process_func` function to preprocess the testing dataset and remove original columns.
test_dataset = test_ds.map(process_func, remove_columns=test_ds.column_names)
```

```python
# Save the preprocessed training dataset to a new JSONL file.
with open('cooked-dataset/tokenized-train.jsonl', "w", encoding="utf-8") as file:
    for message in train_dataset:
        file.write(json.dumps(message, ensure_ascii=False) + "\n")
```

```python
# Save the preprocessed testing dataset to a new JSONL file.
with open('cooked-dataset/tokenized-test.jsonl', "w", encoding="utf-8") as file:
    for message in test_dataset:
        file.write(json.dumps(message, ensure_ascii=False) + "\n")
```

```python
import json
import pandas as pd
import torch
from datasets import Dataset
from peft import LoraConfig, TaskType, get_peft_model
from transformers import AutoModelForCausalLM, TrainingArguments, Trainer, DataCollatorForSeq2Seq, AutoTokenizer
import os
from sklearn.model_selection import train_test_split
from tqdm import tqdm
from peft import PeftModel
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define the directory where model checkpoints are stored.
ckpts_dir = "output\Qwen1.5"

# List all checkpoint files in the directory that contain 'checkpoint-' in their names.
ckpts = [os.path.join(ckpts_dir, f) for f in os.listdir(ckpts_dir) if 'checkpoint-' in f]

# Sort the checkpoint files in ascending order.
ckpts = sorted(ckpts)

# Define the directory where the pre-trained Qwen model is stored.
qwen_dir = "Qwen\Qwen2___5-0___5B-Instruct"
```

```python
# Define the LoRA configuration for fine-tuning.
config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,  # Specify the task type as causal language modeling.
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],  # Target specific layers for LoRA adaptation.
    inference_mode=False,  # Set to False for training mode.
    r=8,  # Rank of the low-rank matrices.
    lora_alpha=32,  # Scaling factor for the low-rank matrices.
    lora_dropout=0.1,  # Dropout rate for LoRA layers.
)
```

```python
# Define a function to load the tokenizer and model with LoRA adaptation.
def load_models(qwen_dir, ckpt):
    # Load the tokenizer from the pre-trained model directory.
    tokenizer = AutoTokenizer.from_pretrained(qwen_dir)

    # Load the pre-trained causal language model.
    model = AutoModelForCausalLM.from_pretrained(qwen_dir)

    # Apply LoRA adaptation to the model using the specified checkpoint.
    model = PeftModel.from_pretrained(model, model_id=ckpt, config=config)

    # Set the model to evaluation mode.
    model.eval()

    # Return the tokenizer and the adapted model.
    return tokenizer, model
```

```python
# Load the test dataset from a JSONL file.
test_dataset = pd.read_json('cooked-dataset/test.jsonl', lines=True)
```

```python
# Define a function to preprocess the test dataset for model evaluation.
def preprocess(tokenizer, test_dataset):
    # Initialize empty lists to store processed inputs and ground truth labels.
    processed_list = []
    ground_truths = []

    # Create a progress bar to track the preprocessing progress.
    iterator = tqdm(range(test_dataset.shape[0]))

    # Iterate over each row in the test dataset.
    for line in iterator:
        # Extract the current row from the dataset.
        line = test_dataset.iloc[line]

        # Format the input as a chat message with system and user roles.
        msg = [
            {'role': 'system', 'content': line['instruction']},  # System instruction.
            {'role': 'user', 'content': line['input']}  # User input.
        ]

        # Apply the tokenizer's chat template to format the message.
        # `tokenize=False` returns the formatted text as a string.
        # `add_generation_prompt=True` adds a prompt for the assistant's response.
        text = tokenizer.apply_chat_template(msg, tokenize=False, add_generation_prompt=True)

        # Tokenize the formatted text and convert it into PyTorch tensors.
        model_inputs = tokenizer([text], return_tensors="pt")

        # Append the tokenized inputs to the processed list.
        processed_list.append(model_inputs)

        # Append the ground truth label (expected output) to the list.
        ground_truths.append(line['output'])

    # Return the list of processed inputs and ground truth labels.
    return processed_list, ground_truths
```

```python
tokenizer, model =  load_models(qwen_dir, ckpts[0])
test_dataset, ground_truths = preprocess(tokenizer, test_dataset)
```

```python
@torch.no_grad()
def test_accuracy(model, tokenizer, test_dataset, ground_truths):
    predictions = []
    truths = []
    iterator = tqdm(range(len(test_dataset)))
    for idx in iterator:
        model_inputs_1 = test_dataset[idx].to('cuda')

        generated_ids_1 = model.generate(
            model_inputs_1.input_ids,
            max_new_tokens=512
        )

        generated_ids_1 = [
            output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs_1.input_ids, generated_ids_1)
        ]

        response_1 = tokenizer.batch_decode(generated_ids_1, skip_special_tokens=True)[0]

        predictions.append(response_1)

        truths.append(ground_truths[idx])
        iterator.set_postfix({
            'Accuracy': accuracy_score(truths, predictions)
        })
    return accuracy_score(truths, predictions)
```

```python
metric_list = []
ckpt_names = []
for ckpt in ckpts[:-1]: # load ckpt path
    for i in range(5): # ensemble k models
        tokenizer, model = load_models(qwen_dir, ckpt) # load model
        model.cuda()
        acc = test_accuracy(model, tokenizer, test_dataset, ground_truths) # get test accuracy
        # save acc
        metric_list.append(acc)
        ckpt_names.append(int(ckpt.split('-')[-1]))
        try:
            # plot
            # scatter plot
            plt.figure(figsize=(20, 5))
            colors = plt.cm.jet(np.linspace(0, 1, len(ckpt_names)))
            plt.scatter(ckpt_names, metric_list, marker='o', s=50, alpha=0.6, edgecolors='w')
            plt.xlabel('Steps')
            plt.ylabel('Accuracy')
            plt.title('Accuracy over Steps')
            plt.grid(True)
            plt.tight_layout()
            plt.savefig('./scatter_plot_5shot.png')
            plt.close()
            # boxplot
            df = {}
            for ckpt_name in ckpt_names:
                df[ckpt_name] = []
            for ckpt_name, metric in zip(ckpt_names, metric_list):
                df[ckpt_name].append(metric)
            df = pd.DataFrame(df)
            df.columns = ckpt_names
            plt.figure(figsize=(20, 6))
            df.boxplot()
            plt.title('Boxplot of Metrics')
            plt.xlabel('Metrics')
            plt.grid(True)
            plt.tight_layout()
            plt.savefig('./box_plot_5shot.png')
            plt.close()
        except:
            continue
```
